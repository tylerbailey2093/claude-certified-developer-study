# Lab 08 · An MCP server with all three primitives
**~30 minutes · costs nothing · Domain 8 MCP Development**

No API calls. You write a server, drive it over stdio with the SDK's own client,
then connect it to Claude Code or Claude Desktop. Knowing **who controls each
primitive** is the testable distinction:

| Primitive | Controlled by | Purpose |
|---|---|---|
| Tools | **Model** | Actions it can invoke |
| Resources | **Application** | Read-only data the client attaches as context |
| Prompts | **User** | Reusable templates the user invokes, often as slash commands |

Written against the Python SDK `mcp` 2.x. In 2.x `FastMCP` was renamed
`MCPServer` (`from mcp.server.mcpserver import MCPServer`); importing
`mcp.server.fastmcp` now raises `ModuleNotFoundError`. Old tutorials will break
on that line. The decorators did not change.

## Write the server

````python
server = '''from mcp.server.mcpserver import MCPServer
from mcp.server.mcpserver.exceptions import ToolError

mcp = MCPServer("warehouse")

STOCK = {"WID-1": 42, "WID-2": 0, "WID-3": 118}

@mcp.tool()
def check_stock(sku: str) -> int:
    """Return the current stock level for a SKU. Use when asked whether an
    item is available or how many remain. Do NOT use to place orders."""
    if sku not in STOCK:
        # ToolError: the model sees this message and can correct itself.
        raise ToolError(f"Unknown SKU {sku!r}. Known SKUs: {sorted(STOCK)}")
    return STOCK[sku]

@mcp.resource("catalog://all")
def catalog() -> str:
    """The full SKU catalogue as read-only reference data."""
    return "\\n".join(f"{k}: {v} units" for k, v in STOCK.items())

@mcp.prompt()
def restock_review(sku: str) -> str:
    """Template for a restocking decision."""
    return f"Review stock for {sku} and recommend whether to reorder."

if __name__ == "__main__":
    mcp.run()          # stdio: the client spawns this process
'''
open("warehouse_server.py", "w").write(server)
print("wrote warehouse_server.py")
````

## Sanity check it starts

stdio servers speak newline-delimited JSON-RPC on stdin/stdout, so running one
bare just waits for input. That is correct behaviour, not a hang.

````python
import subprocess, sys, time
p = subprocess.Popen([sys.executable, "warehouse_server.py"],
                     stdin=subprocess.PIPE, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(2)
print("still running (expected):", p.poll() is None)
p.terminate(); p.wait(timeout=5)
err = p.stderr.read().decode()[:400]
print("stderr:", err if err else "(clean)")
````

## Drive it the way a host does

A host (Claude Code, Claude Desktop) is an MCP client: it **launches the server
as a subprocess**, discovers the three primitives, and calls them. The SDK's
`Client` does exactly that. It runs as a separate script because the notebook
already owns an event loop.

````python
client_src = '''import asyncio, sys
from mcp import Client, StdioServerParameters

def text(result):
    return " ".join(b.text for b in result.content if b.type == "text")

async def main():
    server = StdioServerParameters(command=sys.executable, args=["warehouse_server.py"])
    async with Client(server) as c:
        print("protocol :", c.protocol_version)
        print("tools    :", [t.name for t in (await c.list_tools()).tools])
        print("resources:", [str(r.uri) for r in (await c.list_resources()).resources])
        print("prompts  :", [p.name for p in (await c.list_prompts()).prompts])
        print()
        tool = (await c.list_tools()).tools[0]
        print("tool schema the model sees:", tool.input_schema)
        r = await c.call_tool("check_stock", {"sku": "WID-3"})
        print("call_tool(WID-3) ->", text(r), "| is_error:", r.is_error)
        r = await c.call_tool("check_stock", {"sku": "WID-9"})
        print("call_tool(WID-9) ->", text(r), "| is_error:", r.is_error)
        r = await c.read_resource("catalog://all")
        print("read_resource    ->", repr(r.contents[0].text))
        r = await c.get_prompt("restock_review", {"sku": "WID-2"})
        print("get_prompt       ->", r.messages[0].role, "|", r.messages[0].content.text)

asyncio.run(main())
'''
open("warehouse_client.py", "w").write(client_src)
out = subprocess.run([sys.executable, "warehouse_client.py"], capture_output=True, text=True, timeout=60)
print(out.stdout or out.stderr[-2000:])
````

## Connect it

**Claude Code** (stdio is the default transport; `--` separates Claude's options
from the server command):
```
claude mcp add warehouse -- python /full/path/to/warehouse_server.py
```
Add `--scope project` to write it to `.mcp.json` and share it through version
control. Check status with `claude mcp list` or `/mcp` inside a session.

**Claude Desktop** reads the same information from the `mcpServers` key of
`claude_desktop_config.json`. The next cell prints the entry with this

````python
import json, pathlib
desktop = {"mcpServers": {"warehouse": {
    "command": sys.executable,
    "args": [str(pathlib.Path("warehouse_server.py").resolve())]}}}
print(json.dumps(desktop, indent=2))
````

Restart the host, then ask *"how many WID-3 do we have?"* and watch the tool
get called.

### The transport heuristic
Both configs give a **command to run**: that is **stdio**. The client
launches the server as a local subprocess; nothing listens on the network. If a
server's docs give you a **URL** instead, that is **Streamable HTTP**, the
standard remote transport (`claude mcp add --transport http <name> <url>`).
Remote servers should be authenticated; the spec's OAuth-based authorization is
optional for HTTP and not used for stdio. The older HTTP+SSE transport has been
deprecated since protocol revision 2025-03-26.
