## The Models API

Blueprint-relevant: query limits rather than hardcoding them. `max_input_tokens`
and `max_tokens` are two different ceilings and the exam tests that you know it.

````python
for m in client.models.list():
    print(f"{m.id:<34} in={getattr(m,'max_input_tokens','?'):<9} out={getattr(m,'max_tokens','?')}")
````

**Set your working models from the list above, then run the rest of the labs.**

````python
MODEL = "claude-sonnet-4-6"     # <- edit if the list above disagrees
CHEAP = "claude-haiku-4-5"

r = client.messages.create(model=MODEL, max_tokens=64,
        messages=[{"role":"user","content":"Reply with exactly: ok"}])
print(r.content[0].text, "|", r.stop_reason, "|", r.usage.input_tokens, "in /", r.usage.output_tokens, "out")
````
