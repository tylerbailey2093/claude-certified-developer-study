# Lab 09 · A hook that actually blocks
**~20 minutes · costs nothing · Domain 7 Hooks + Security, Domain 3**

The load-bearing idea in three domains: **a prompt instruction is a request, a
hook is a control.** And the detail that catches people: **exit 2 blocks; any
other non-zero exit is a non-blocking error and the action proceeds.** A
crashing guard script lets the command through.

````python
import os, json, pathlib, subprocess, sys
root = pathlib.Path("hooklab"); (root/".claude"/"hooks").mkdir(parents=True, exist_ok=True)
print(root.resolve())
````

## The guard script

````python
guard = r'''#!/usr/bin/env bash
# Reads the hook payload on stdin. No jq dependency: parse with python3.
input=$(cat)
command=$(printf '%s' "$input" | python3 -c 'import sys,json; print(json.load(sys.stdin).get("tool_input",{}).get("command",""))' 2>/dev/null)

if printf '%s' "$command" | grep -qiE '(^|[[:space:]])rm[[:space:]]+(-[a-z]*r|--recursive)'; then
  echo "Recursive rm is blocked by project policy." >&2
  exit 2      # BLOCKS. stderr is returned to Claude as the reason.
fi
exit 0        # proceeds
'''
p = root/".claude"/"hooks"/"guard.sh"
p.write_text(guard); p.chmod(0o755)

settings = {"hooks": {"PreToolUse": [{"matcher": "Bash",
  "hooks": [{"type": "command", "command": "${CLAUDE_PROJECT_DIR}/.claude/hooks/guard.sh"}]}]}}
(root/".claude"/"settings.json").write_text(json.dumps(settings, indent=2))
print(json.dumps(settings, indent=2))
````

## Test the guard directly

Feed it the same JSON shape Claude Code sends on stdin for `PreToolUse`, and
read the **exit code**. That is the entire contract.

````python
def fire(cmd, script=p):
    payload = json.dumps({"session_id": "lab", "cwd": str(root.resolve()),
                          "hook_event_name": "PreToolUse", "tool_name": "Bash",
                          "tool_input": {"command": cmd}})
    r = subprocess.run([str(script)], input=payload, capture_output=True, text=True)
    verdict = {0: "ALLOW", 2: "BLOCK"}.get(r.returncode, "NON-BLOCKING ERROR -> ALLOWED")
    print(f"{cmd:<26} exit={r.returncode:<4} {verdict:<6} {r.stderr.strip()}")

for c in ["ls -la", "rm -rf /tmp/x", "rm --recursive build", "rm file.txt",
          "git status", "/bin/rm -rf build"]:
    fire(c)
````

Look at the last row. A denylist regex is a control, but a leaky one:
`/bin/rm` slips past `(^|[[:space:]])rm`. Hooks are one layer; permission deny
rules and sandboxing are the others.

## The footgun

Break the script the way a real one breaks: a missing dependency, a typo, an
unset variable. It exits non-zero but **not 2**, so the destructive command
proceeds. Your guard failed **open**.

````python
broken = root/".claude"/"hooks"/"broken.sh"
broken.write_text("#!/usr/bin/env bash\ncat > /dev/null\nnosuchcommand_typo\n")
broken.chmod(0o755)

r = subprocess.run([str(broken)],
        input=json.dumps({"tool_name": "Bash", "tool_input": {"command": "rm -rf /"}}),
        capture_output=True, text=True)
print("exit code:", r.returncode, "->", "BLOCKED" if r.returncode == 2 else "ALLOWED")
print()
print("Exit 1 (or 127) is a non-blocking hook error. The tool call still runs.")
print("Guard scripts need their own error handling, or they protect nothing.")
````

## Fail closed instead

For a security guard the safe default is the opposite: if the guard itself
breaks, block. `set -euo pipefail` plus an `ERR` trap that exits 2 turns any
unexpected failure into a block.

````python
hardened = root/".claude"/"hooks"/"hardened.sh"
hardened.write_text(
    "#!/usr/bin/env bash\n"
    "set -euo pipefail\n"
    "trap 'echo \"guard crashed; failing closed\" >&2; exit 2' ERR\n"
    "cat > /dev/null\n"
    "nosuchcommand_typo\n"
    "exit 0\n")
hardened.chmod(0o755)

fire("ls -la", script=hardened)
print()
print("Same typo, opposite outcome. The cost: every Bash call is blocked until")
print("someone fixes the guard. For a security control that is the right trade.")
````
