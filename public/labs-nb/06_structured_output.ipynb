# Lab 06 · Schemas that let the model decline
**~25 minutes · costs about $0.02 · Domain 2 App Design (8.6%), Domain 6 Output Handling**

The heaviest objective on the exam includes schema design. One design choice
does most of the work: **nullable fields**. If a schema cannot express *"not
present"*, the model must put something there.

This lab uses structured outputs (`output_config.format`), so the JSON shape is
enforced by the API, not requested in a prompt. That makes the lesson sharper:
a guaranteed shape still cannot stop fabrication when the shape has no room for
"unknown".

```python
import os, anthropic
client = anthropic.Anthropic()          # reads ANTHROPIC_API_KEY
MODEL  = "claude-sonnet-5-5"            # everyday model (src/data/facts.json)
CHEAP  = "claude-haiku-4-5"             # cheap, high-volume calls

def text_of(resp):
    """Join the text blocks. Never index content[0]: it can be a thinking block."""
    return "".join(b.text for b in resp.content if b.type == "text")

print("sdk", anthropic.__version__)
```

## A contract with a genuinely missing field

```python
CONTRACT = """MASTER SERVICES AGREEMENT
Vendor: Northwind Logistics Ltd.
Effective: 3 March 2026.
This agreement renews automatically and continues until terminated by either
party with 90 days written notice. Payment terms are net 45."""

import json
def extract(schema, label, model=MODEL):
    # max_tokens is generous on purpose: thinking on Sonnet 5.5 counts toward it.
    r = client.messages.create(
        model=model, max_tokens=4096,
        system="Extract the requested fields from the contract.",
        messages=[{"role": "user", "content": CONTRACT}],
        output_config={"format": {"type": "json_schema", "schema": schema}},
    )
    print(f"{label:<18} stop_reason={r.stop_reason}")
    print("   ", text_of(r))
    return text_of(r)
```

## Version A · no way to say "unknown"

There is **no end date** in this contract; it runs until terminated. With
`output_config.format` the response must match the schema, so for this field the
model can only emit a string. Watch which string it picks.

```python
A = {"type": "object",
     "properties": {
        "vendor": {"type": "string"},
        "contract_end_date": {"type": "string", "description": "ISO date"}},
     "required": ["vendor", "contract_end_date"],
     "additionalProperties": False}
raw_a = extract(A, "A (not nullable)")
```

## Version B · nullable, with the reason spelled out

Same model, same contract. The only change is that declining is now
representable.

```python
B = {"type": "object",
     "properties": {
        "vendor": {"type": "string"},
        "contract_end_date": {"type": ["string", "null"],
            "description": "ISO date, or null if the contract has no fixed end date"}},
     "required": ["vendor", "contract_end_date"],
     "additionalProperties": False}
raw_b = extract(B, "B (nullable)")
```

Run both a few times. Under A the only legal outputs are strings, so you get
either a plausible invented date or a free-text placeholder ("N/A", "until
terminated") that breaks every downstream date parser. Under B, `null` is legal
and is what comes back. Marking a field *required* without making it nullable
makes fabrication **more** likely, not less: it removes the escape hatch while
insisting on a value.

## Validation is not the same as correctness

```python
def validate(raw):
    try:
        d = json.loads(raw)
    except json.JSONDecodeError as e:
        return f"INVALID JSON: {e}"
    if "vendor" not in d:
        return "SCHEMA FAIL: missing vendor"
    if d.get("contract_end_date") and not str(d["contract_end_date"])[:4].isdigit():
        return "SEMANTIC FAIL: end date is not a date"
    return f"passes both: {d}"

print(validate('{"vendor":"Northwind Logistics Ltd.","contract_end_date":null}'))
print(validate('{"vendor":"Northwind Logistics Ltd.","contract_end_date":"until terminated"}'))
print(validate('{"vendor":"Northwind Logistics Ltd.","contract_end_date":"2031-03-03"}'))
print("   ^ schema-valid and completely fabricated. Validation cannot catch this.")
```

Structured outputs make the first two branches unreachable on a complete
response. The third failure is the one left, and no validator can see it: the
value is well-formed and wrong. Only the schema design (B) prevents it.

## The truncation trap

The structured-outputs guarantee has an exception the docs state plainly: if
the response hits `max_tokens`, the output may be incomplete and not match the
schema. Force it. The instinct is to send the fragment back for repair. Do not.

```python
WIDE = {"type": "object",
        "properties": {k: {"type": "string"} for k in [
            "vendor", "effective_date", "renewal", "termination", "notice_period",
            "payment_terms", "governing_law", "liability_cap", "currency",
            "signatories", "services", "summary"]},
        "required": ["vendor", "effective_date", "renewal", "termination", "notice_period",
                     "payment_terms", "governing_law", "liability_cap", "currency",
                     "signatories", "services", "summary"],
        "additionalProperties": False}

# CHEAP runs without thinking by default, so all 40 tokens go to the JSON.
r = client.messages.create(model=CHEAP, max_tokens=40,
        system="Describe the contract in detail.",
        messages=[{"role": "user", "content": CONTRACT}],
        output_config={"format": {"type": "json_schema", "schema": WIDE}})
print("stop_reason:", r.stop_reason)
print("payload:", repr(text_of(r)))
print()
if r.stop_reason == "max_tokens":
    print("Detect this BEFORE json.loads. Raise the budget or decompose the task.")
    print("A repair loop asks the model to invent data that was never generated.")
```
