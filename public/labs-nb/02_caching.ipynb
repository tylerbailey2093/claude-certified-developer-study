# Lab 02 · Prompt caching, and how to destroy it
**~25 minutes · costs about $0.05 · Domain 2 API Mechanics, Domain 5 Cost**

The single highest-value lab here. Most caching cost problems come down to one
fact: **caching is an exact prefix match**. You will see your own hit rate
collapse because of one interpolated value, and you will not forget it
afterwards.

```python
import os, anthropic
client = anthropic.Anthropic()          # reads ANTHROPIC_API_KEY
MODEL  = "claude-sonnet-5-5"            # everyday model (src/data/facts.json); verify against lab 00
CHEAP  = "claude-haiku-4-5"             # for high-volume steps

def text_of(resp):
    """Join every text block. Never index content[0]: the first block can be thinking or tool_use."""
    return "".join(b.text for b in resp.content if b.type == "text")

print("sdk", anthropic.__version__)
```

## Build a prefix big enough to cache

The minimum cacheable prefix is model-dependent: 512 tokens on Sonnet 5.5 and
Opus 5.5, 1,024 on Sonnet 5 and Opus 5, and 4,096 on Haiku 4.5 (prompt caching
docs, checked 2026-10-06). Below the minimum nothing errors; the request just
runs uncached with `cache_creation_input_tokens: 0`. Note that the cheapest
model needs the largest prefix, so a short system prompt that caches on Sonnet
can silently never cache on Haiku.

The corpus below is sized to clear every current minimum.

```python
DOC = ("Section %d. The widget subsystem handles inventory reconciliation "
       "across regional warehouses, applying tier rules and audit logging.\n")
CORPUS = "".join(DOC % i for i in range(1, 201))
print("approx tokens:",
      client.messages.count_tokens(model=MODEL,
          messages=[{"role":"user","content":CORPUS}]).input_tokens)
```

## Run 1 · cache write

The first call pays a write premium: 1.25x base input at the default 5-minute
TTL (2x if you ask for the 1-hour TTL).

```python
def ask(question, prefix):
    r = client.messages.create(
        model=MODEL, max_tokens=1024,
        system=[
            {"type":"text","text":"You answer from the provided corpus only."},
            {"type":"text","text":prefix,"cache_control":{"type":"ephemeral"}},
        ],
        messages=[{"role":"user","content":question}],
    )
    u = r.usage
    print(f"  write={u.cache_creation_input_tokens:<7} read={u.cache_read_input_tokens:<7} "
          f"uncached_in={u.input_tokens}")
    return r

print("run 1"); ask("What does the widget subsystem do?", CORPUS);
```

## Run 2 · cache read

Same prefix. `write` should drop to 0 and `read` should jump. Only the
question after the breakpoint is billed as uncached input.

```python
print("run 2"); ask("Which rules does it apply?", CORPUS);
print("run 3"); ask("Does it log?", CORPUS);
```

## Now break it

One timestamp at the top of the prefix. Nothing else changes.

This is a typical failure scenario: *a release adds a current timestamp to the
cached system prompt, and `cache_read_input_tokens` drops to near zero while
nothing errors.*

```python
from datetime import datetime
print("poisoned run 1"); ask("What does it do?", f"Generated: {datetime.now()}\n" + CORPUS);
print("poisoned run 2"); ask("Which rules?",     f"Generated: {datetime.now()}\n" + CORPUS);
print()
print("read stays at 0. Every call is now a fresh write at 1.25x input price.")
print("The fix is PLACEMENT, not TTL: move volatile values out of the cached prefix.")
```

## Do the arithmetic

Cache reads cost 0.1x base input on most models (0.05x on Opus 5.5, 0.025x on
Fable 5.1). The numbers below use Sonnet 5.5 list prices ($2 per million input
tokens, pricing page checked 2026-10-06). Work out what the mistake costs at
your real traffic volume, then notice the third line: a poisoned cache is
worse than no cache at all.

```python
IN_PER_MTOK, CACHE_READ_MULT, CACHE_WRITE_MULT = 2.00, 0.10, 1.25
tokens, calls = 5000, 100_000

full     = tokens/1e6 * IN_PER_MTOK * calls
cached   = (tokens/1e6*IN_PER_MTOK*CACHE_WRITE_MULT) + (tokens/1e6*IN_PER_MTOK*CACHE_READ_MULT*(calls-1))
poisoned = tokens/1e6 * IN_PER_MTOK * CACHE_WRITE_MULT * calls
print(f"no cache      ${full:,.2f}")
print(f"cached        ${cached:,.2f}")
print(f"poisoned      ${poisoned:,.2f}  (every call a write)")
print(f"saved         ${full-cached:,.2f}  ({(1-cached/full)*100:.1f}%)")
```
