# Lab 07 · Token counting and why migrations need re-measuring
**~20 minutes · costs about $0.001 (count_tokens is free) · Domain 5 (16.8%)**

Token counts are **model-specific**. They do not transfer across a migration.
Claude Opus 4.7 and later models (Sonnet 5 and Sonnet 5.5 included) use a newer
tokenizer: the docs put the same text at roughly 30% more tokens than on Sonnet
4.6 or Haiku 4.5, depending on content. `count_tokens` is free (rate limited)
and returns an estimate, so measure rather than assume.

```python
import os, anthropic
client = anthropic.Anthropic()          # reads ANTHROPIC_API_KEY
MODEL  = "claude-sonnet-5-5"            # everyday model (src/data/facts.json)
CHEAP  = "claude-haiku-4-5"             # cheap, high-volume calls

def text_of(resp):
    """Join the text blocks. Never index content[0]: it can be a thinking block."""
    return "".join(b.text for b in resp.content if b.type == "text")

print("sdk", anthropic.__version__)
```

## The same text, measured against several models

```python
SAMPLE = """def reconcile(ledger, statements, tolerance=0.01):
    unmatched = []
    for entry in ledger:
        hit = next((s for s in statements if abs(s.amount - entry.amount) < tolerance), None)
        if hit is None:
            unmatched.append(entry)
    return unmatched"""

COMPARE = [CHEAP, "claude-sonnet-4-6", MODEL, "claude-opus-5-5"]
base = None
for mid in COMPARE:
    try:
        n = client.messages.count_tokens(model=mid,
                messages=[{"role": "user", "content": SAMPLE}]).input_tokens
        base = base or n
        print(f"{mid:<22} {n:>6} tokens   {n/base:>5.2f}x")
    except anthropic.APIStatusError as e:
        print(f"{mid:<22} n/a ({e.status_code})")
```

## Count the whole request, not just the message

Tools and the system prompt are input tokens too. This is the number your cost
model needs.

```python
TOOLS = [{"name": "search", "description": "Search the knowledge base for relevant passages.",
          "input_schema": {"type": "object", "properties": {"q": {"type": "string"}}, "required": ["q"]}}]
SYS = "You are a meticulous financial reconciliation assistant." * 8

for label, kw in [
    ("message only",     dict(messages=[{"role": "user", "content": SAMPLE}])),
    ("+ system",         dict(system=SYS, messages=[{"role": "user", "content": SAMPLE}])),
    ("+ system + tools", dict(system=SYS, tools=TOOLS, messages=[{"role": "user", "content": SAMPLE}])),
]:
    print(f"{label:<20}", client.messages.count_tokens(model=MODEL, **kw).input_tokens)
```

## Context window is not max output

Two different ceilings. The exam tests that you know it. The Models API reports
both: `max_input_tokens` is the context window, `max_tokens` is the largest
value you may pass as `max_tokens`.

```python
for m in client.models.list():
    mi, mo = getattr(m, "max_input_tokens", None), getattr(m, "max_tokens", None)
    if mi:
        print(f"{m.id:<24} context={mi:<10} max_output={mo}")
```

## Effort, and the parameters that now return 400

> **Engineering knowledge, not exam content.** These restrictions shipped with
> models released after Exam Guide v1.0 (July 2026), so exam items were almost
> certainly not written against them. (Prefill rejection started earlier, with
> the 4.6 family; the rest arrived later.) Know them so your code runs; do not
> build exam answers on them.

On the current Sonnet the request-level controls changed:

- non-default `temperature`, `top_p` or `top_k`: **400**
- `thinking: {"type": "enabled", "budget_tokens": N}`: **400** (adaptive thinking plus effort replaces it)
- `thinking: {"type": "disabled"}`: **400** on Sonnet 5.5 (its off switch is `{"type": "between_tools"}`)
- forced `tool_choice` (`any` or `tool`): **400** (use `auto` plus `strict: true`, or structured outputs)
- a prefilled final assistant turn: **400** (use structured outputs or instructions)

Two layers can reject a request. The Python SDK 1.x no longer has a
`temperature` keyword at all, so passing one raises `TypeError` before anything
is sent. Sending it through `extra_body` gets it to the server, which returns
the 400.

`output_config.effort` is the control surface now. It is a behavioral signal,
not a token cap; `max_tokens` is still the only hard ceiling. The cell below
catches each error so the notebook keeps running. Confirm it yourself rather
than trusting this list.

```python
Q = [{"role": "user", "content": "What is 17*23? Answer with the number only."}]
CALC = [{"name": "calc", "description": "Evaluate an arithmetic expression.",
         "input_schema": {"type": "object", "properties": {"expr": {"type": "string"}}, "required": ["expr"]}}]

ATTEMPTS = [
    ("effort=low (control)",   dict(output_config={"effort": "low"})),
    ("thinking between_tools", dict(thinking={"type": "between_tools"})),
    ("temperature kwarg",      dict(temperature=0.2)),
    ("temperature extra_body", dict(extra_body={"temperature": 0.2})),
    ("budget_tokens=2000",     dict(thinking={"type": "enabled", "budget_tokens": 2000})),
    ("thinking disabled",      dict(thinking={"type": "disabled"})),
    ("tool_choice any",        dict(tools=CALC, tool_choice={"type": "any"})),
    ("assistant prefill",      dict(messages=Q + [{"role": "assistant", "content": "The answer is"}])),
]

def api_msg(e):
    """The server's own message. e.message is prefixed with "Error code: 400 - {...}"."""
    body = e.body if isinstance(e.body, dict) else {}
    return str(body.get("error", {}).get("message", e.message))[:100]

for label, kw in ATTEMPTS:
    try:
        r = client.messages.create(model=MODEL, max_tokens=2048, **({"messages": Q} | kw))
        print(f"{label:<24} OK   out={r.usage.output_tokens:<5} {text_of(r)[:40]!r}")
    except anthropic.BadRequestError as e:          # the 400 this cell exists to show
        print(f"{label:<24} 400  {api_msg(e)}")
    except anthropic.APIStatusError as e:
        print(f"{label:<24} {e.status_code}  {api_msg(e)}")
    except anthropic.APIConnectionError as e:
        print(f"{label:<24} no connection: {type(e).__name__}")
    except TypeError as e:
        print(f"{label:<24} SDK rejected before sending: {e}")
```
